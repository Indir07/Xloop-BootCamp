```python
print("Welcome to Campus Cafe")

total_bill = 0
discount = 0


name = input("Enter customer name: ")

name = name.replace(" ", "")

if name[0].isupper() and name.isalpha():
    print("\nManu:\n 1. Tea RS. 100\n", "2. Sandwich RS. 250\n", "3. Burger RS. 450\n")

    while True:
        choice = input("Enter (1/2/3) or enter (done) to finish order: ")

        if choice == "done":
            break
        elif choice == "1" or choice == "2" or choice == "3":
            q = input("Enter quantity: ")

            if q.isdigit() and int(q) > 0:
                q = int(q)
                if choice == "1":
                    total_bill += 100 * q
                elif choice == "2":
                    total_bill += 250 * q
                else:
                    total_bill += 450 * q
            else:
                print("Invalid quatity")
                break
        else:
            print("Invalid choice of manu")
            break

    if total_bill >= 1000:
        discount = 0.1 * total_bill

    total_bill = total_bill - discount

    print("--------------------------------")
    print("Campus Cafe")
    print("Customer Name:", name)
    print("Total bill: RS.", total_bill)
    print("Discount: RS.", discount)
    print("Final bill: RS.", total_bill)
    print("Thank you for order")
    print("--------------------------------")
else:
    print("First character of name must be capital")
```

```python
print("Electricity Billing System")

while True:
    total_bill = 0

    for i in range(1, 6):
        print(f"\nCustomer {i}")

        name = input("Enter customer name: ").title()
        units = int(input("Enter units consumed: "))

        if units <= 100:
            e_bill = units * 10
        elif units <= 200:
            e_bill = 1000 + (units - 100) * 15
        else:
            e_bill = 2500 + (units - 200) * 20

        bill = e_bill + 200

        surcharge = 0

        if units > 200 and bill >= 3000:
            surcharge = bill * 0.05
            bill += surcharge

        total_bill += bill

        print(f"\nCustomer: {name}")
        print(f"Units consumed: {units}")
        print(f"Energy charge: Rs. {e_bill}")
        print("Service charge: Rs. 200.00")
        print(f"Surcharge: Rs. {surcharge}")
        print(f"Total bill: Rs. {bill}")

    print(f"\nTotal revenue: Rs. {total_bill}")

    a = input("\nContinue new batch? (yes/no): ")
    if a == "no":
        print("Billing ended")
        break
```

```python
a_rooms = 5
total_cost = 0
r_rooms = 0
revenue = 0

while True:
    name = input("Enter guest name: ").strip().title()
    room_type = input("Enter room type (Standard/Deluxe): ").strip().lower()
    n_rooms = int(input("Enter number rooms: "))
    n_nights = int(input("Enter number of nights: "))
    m_status = input("Do you have membership (yes/no)").strip().lower()

    a_rooms -= n_rooms
    r_rooms += n_rooms

    if a_rooms >= 0:
        # print("Rooms available")
        if room_type == "deluxe":
            total_cost = n_rooms * 5000 * n_nights
        elif room_type == "standard":
            total_cost = n_rooms * 3000 * n_nights
        else:
            print("Invalid room type")
            break
    else:
        print("Room not availabe")
        break

    total_cost = total_cost * n_rooms

    final_amount = 0
    n_discount = 0

    if n_nights >= 3:
        n_discount = total_cost * 0.1

    m_discount = 0
    if m_status == "yes":
        m_discount = total_cost * 0.05

    final_amount = total_cost - n_discount - m_discount
    revenue += final_amount

    print("\n" + "-" * 25)
    print(f"Guset Name: {name}")
    print(f"Room type: {room_type}")
    print(f"Number of rooms: {n_rooms}")
    print(f"Number of nights: {n_rooms}")
    print(f"Original bill: {total_cost}")
    print(f"Nights discount: {n_discount}")
    print(f"Membership discount: {m_discount}")
    print(f"Final Amount: {final_amount}")
    print("-" * 25 + "\n")
    print(f"Total reserved rooms: {(r_rooms)}")
    print(f"Remaining rooms: {(a_rooms)}")
    if a_rooms == 0:
        break
print(f"Total expected revenue: {revenue}")
```
