1. Phone Number Validation Program

```python
phone = input("Enter phone number: ")
print(phone)

length = len(phone)  

# Format 1 0332-3749390  (length = 12)
if length == 12 and phone[0] == '0' and phone[1] == '3' and phone[4] == '-' and phone[2:4].isdigit() and phone[5:].isdigit():
    print("Valid phone number")

# Format 2: 03323749390  (length = 11)  
elif length == 11 and phone[0] == '0' and phone[1] == '3' and phone.isdigit():
    print("Valid phone number")
     

# Format 3: +923323749390  (length = 13)
elif length == 13 and phone[0] == '+' and phone[1] == '9' and phone[2] == '2' and phone[4:].isdigit():
    print("Valid phone number ")


else:
    print("Invalid phone number")
```

2. String Validation and Manipulation

```python
email = input("Enter email address: ")

user = email.split('@')
username = user[0]
domain = user[1]

if ' ' in email or email.count('@') == 1 and '.' not in domain or email.startswith(".") or email.endswith(".") or email.startswith("@"):
    print("Invalid email address.")
else:

    modified_domain = domain.replace('.', '-')
    total_char = len(email)
    uppercase_email = email.upper()
    capital_username = username.capitalize()

    print("Username:" ,username)
    print("Domain:", domain)
    print("Modified Domain:", modified_domain)
    print("Total Characters:", total_char)
    print("Uppercase Email:", uppercase_email)
    print("Capitalized Username:", capital_username)
```

4. Multiplication

```python
# Basic table for 2 from 1 to 10
print("Basic Multiplication Table for 2")
for i in range(1, 11):
    p = 2 * i
    print("2 x", i, "=", p)


print("\n")

# Show only odd or even multiples
u = input("Enter Multiples of 2 (Even/Odd)")
if u == "Even":
    print("Even Multiples of 2")
    for i in range(1, 11):
        p = 2 * i
        if i % 2 == 0:
            print("2 x", i, "=", p)
elif u == "Odd":
    print("Odd Multiples of 2")
    for i in range(1, 11):
        p = 2 * i
        if i % 2 != 0:
            print("2 x", i, "=", p)
else:
    print("Invalid input")


# Dynamic table for user-entered number and limit
print("\nDynamic Multiplication Table")
number = int(input("Enter a number: "))
limit = int(input("Enter the limit: "))

for i in range(1, limit + 1):
    p = number * i
    print(number, "x", i, "=", p)

# Reverse order option
print("\nReverse Order Multiplication Table")
reverse = input("Do you want reverse order? (yes/no): ")
if reverse == "yes":
    for i in range(limit, 0, -1):
        p = number * i
        print(number, "x", i, "=", p)
else:
    print("Table is in normal order.")
```
