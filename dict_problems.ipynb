# Problem 1: Student Performance Record


```python
students = {"Ali": 85, "Sara": 92, "Ahmed": 76, "Fatima": 89, "Usman": 68}


print(students.keys())
print(students.values())

print("")
for key, value in students.items():
    print(f"Name: {key}, Marks: {value}")

students.update({"Hina": 95})
print(students)

students.update({"Ahmed": 82})
print(students)

print(students.get("Sara", "Student not found"))

print(f"\nNumber of students: {len(students)}")

print("Students who have 80 marks or more")
for key, value in students.items():
    if value >= 80:
        print(f"Name: {key}")

print("\nClass Average: ", sum(students.values()) / len(students))

print("\n", students.get("Bilal", 0))
```

# Problem 2: Grocery Store Stock and Restocking


```python
stock = {"Rice": 25, "Sugar": 15, "Flour": 30, "Milk": 10, "Eggs": 40}


for key, value in stock.items():
    print(f"item: {key}, Quantity: {value}")

print("")
stock["Rice"] = stock["Rice"] - 4
print(stock["Rice"])

print("")
stock["Sugar"] = stock["Sugar"] + 12
print(stock["Sugar"])

stock.update({"Tea": 20})

print(stock.get("Butter", "Product not found"))

print("\nProducts fewer than 15")
for key, value in stock.items():
    if value < 15:
        print(key)

print(sum(stock.values()))

stock.pop("Milk")

print(stock)

choice = input("Enter product name: ")
print(stock.get(choice, 0))
```

# Problem 3: Class Poll Results


```python
votes = {"Sports Day": 12, "Movie Day": 18, "Picnic": 15, "Board Games": 9}

print("Activities")
for key, value in votes.items():
    print(f"Activity: {key}")

print("Activities and their votes count")
for key, value in votes.items():
    print(f"Activity: {key}, Votes: {value}")


votes["Picnic"] = votes["Picnic"] + 3

votes.update({"Talent Show": 0})

print("\n Movies day: ", votes.get("Movie Day", 0))


print("\n ", sum(votes.values()))

print("\nVotes at leat 15:")
for key, value in votes.items():
    if value >= 15:
        print(f"Activity: {key}, Votes: {value}")

for key, value in votes.items():
    if max(votes.values()) == value:
        print(f"Highest votes activity: {key}, Votes: {value}")

print()
print(votes.get("Camping", "Camping not found"))
```

# Problem 4: Contact Management System


```python
print("Contact Management System")

print("\nManu: ")
print(
    "1. Add a new contact	\n2. Search for a contact \n3. Delete a contact \n4. Display all contacts \n5.Update a contact \6. Count total contacts \n7. Display contacts starting with a specific letter \n8. Exit "
)

contacts = {}

while True:
    choice = int(input("Enter you manu choice: "))

    if choice == 8:
        break
    elif choice == 1:
        name = input("Enter name: ")
        phone = input("Enter phone: ")

        if name in contacts:
            ask = input(
                "name already exists, do you want to replace the existing number (yes/no)"
            )

            if ask == "yes":
                contacts.update({name: phone})
                print(f"contact replaced \n {name}, {phone}")
            else:
                continue
        else:
            contacts.update({name: phone})
            print(f"contact added {name}, {phone}")
    elif choice == 2:
        name = input("Enter name you want to search: ")
        print(contacts.get(name, "Contact not fount"))
    elif choice == 3:
        name = input("Enter contact name you want to delete: ")
        if name in contacts:
            ask = input("Delete Confirm (yes/no)")
            if ask == "yes":
                contacts.pop(name)
                print(f"{name} contact deleted")
            else:
                continue
        else:
            print("Contact not fount")
    elif choice == 4:
        for key, value in contacts.items():
            print(f"Name: {key}, phone: {value}")

    elif choice == 5:
        name = input("Enter contact name to update: ")
        if name in contacts:
            new_number = input("Enter new phone number: ")
            contacts[name] = new_number
            print(f"Contact {name} updated.")
        else:
            print("Contact not found.")

    elif choice == 6:
        print(f"Total number of contacts: {len(contacts)}")

    elif choice == 7:
        char = input("Enter letter of contact you want to find: ").strip().lower()
        for key, value in contacts.items():
            if key[0].lower() == char:
                print(f"Name: {key}, phone: {value}")
            else:
                continue
```

# Problem 5: Library Management System


```python
library = {"The Great Gatsby": 5, "1984": 3, "To Kill a Mockingbird": 4, "Moby Dick": 2}

print("Manu:")

print(
    "1. View Available Books \n2. Borrow a Book \n3.Return a Book \n4. Add a New Book \n5. Update Book Copies \n6 Exit"
)

while True:
    choice = input("Enter you choice (1-6) or enter 'done' to exit: ")

    if choice == "done":
        break
    elif choice == "1":
        for key, value in library.items():
            print(f"Book: {key}, Quantity: {value}")
    elif choice == "2":
        ask = input("Enter book you want to borrow: ").strip()
        if ask in library:
            if library[ask] > 0:
                print(f"\n{ask} Book Borrowed")
                library.update({ask: library[ask] - 1})
            else:
                print(f"\n{ask} currently unavailable")
        else:
            print(f"\n{ask} Book not found")
    elif choice == "3":
        ask = input("\nEnter return book name: ")

        if ask in library.keys:
            print(f"\n{ask} Book returned")
            library.update({ask: library[ask] + 1})
        else:
            print(f"\n{ask} not found")
    elif choice == "4":
        name = input("Enter book name: ")
        q = int(input("No. of copies: "))
        library.update({name: q})
        print(f"\n{name} added")
        library.update({ask: library[ask] + 1})

    elif choice == "5":
        name = input("Enter book name: ")
        q = int(input("No. of copies: "))
        library.update({name: q})
        print(f"\n{name} updated")
    else:
        print("\nInvalid manu choice")
```
