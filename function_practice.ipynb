```python
def order_summary(*items, discount=0, tax_rate=0.1, **customer_details):
    total = sum(items)

    if discount > 0:
        discount = total * (discount / 100)

    tax = (total - discount) * tax_rate

    final_amount = total + tax - discount

    for key, value in customer_details.items():
        print(f"{key.capitalize()}: {value}")

    print("-----------Order Bill----------")
    print(f"Total bill: {total:.2f}")
    print(f"Discount: {discount}")
    print(f"Tax applied: {tax:.2f}")
    print(f"Final payable amount: {final_amount:.2f}\n")


order_summary(
    50, 100, 200, discount=10, tax_rate=0.08, name="Alice", address="123 Main St"
)
order_summary(80, 120, name="Bob", discount=5, tax_rate=0.05)
```

```python
def student_report(name, *marks, passing_marks=40, **student_details):
    pass_count = 0

    total_marks = sum(marks)
    avg_marks = total_marks / len(marks)

    print("******************")
    for mark in marks:
        if mark >= passing_marks:
            print("Pass")
            pass_count += 1
        else:
            print("Fail")

    print("------------Student Details--------")
    print(f"Name: {name}")
    for key, val in student_details.items():
        print(f"{key}: {val}")
    print(f"Total marks: {total_marks}")
    print(f"Total pass subjects: {pass_count}")
    print(f"Average marks: {avg_marks}\n")


student_report(
    "Ali", 75, 80, 35, 90, passing_marks=40, class_name="Python", roll_no=101
)
student_report("Sara", 85, 92, 78, section="A")
```

```python
def shopping_cart(*prices, discount=0, delivery_fee=200, **order_details):

    total_bill = sum(prices)

    if discount > 0:
        discount = total_bill * (discount / 100)

    if total_bill >= 5000:
        delivery_fee = 0

    final_amount = total_bill - discount + delivery_fee

    print("---------Order Details--------")
    for key, val in order_details.items():
        print(f"{key.capitalize()}: {val}")
    for i in range(len(prices)):
        print(f"Price of item {i + 1}: {prices[i]}")
    print(f"SubTotal: {total_bill}")
    print(f"Discount: {discount}")
    print(f"Delivery Fee: {delivery_fee}")
    print(f"Total payable amount: {final_amount}\n")


shopping_cart(
    1500, 2500, 1200, discount=10, delivery_fee=200, customer="Ali", city="Karachi"
)
shopping_cart(1000, 800, payment_method="Cash on Delivery")
```

```python
def calculate_payroll(
    name, *bonuses, basic_salary=50000, tax_rate=5, **employee_details
):

    total_bonus = sum(bonuses)

    gross_salary = total_bonus + basic_salary

    tax = gross_salary * (tax_rate / 100)

    net_salary = gross_salary - tax

    print("---------Employee Details--------")
    print(f"Name: {name}")
    for key, val in employee_details.items():
        print(f"{key}: {val}")

    print(f"Basic salary: {basic_salary}")
    print(f"Total Bonus: {total_bonus}")
    print(f"Gross salary: {gross_salary}")
    print(f"Tax: {tax}")
    print(f"Net salary: {net_salary}\n")


calculate_payroll(
    "Ahmed",
    5000,
    3000,
    2000,
    basic_salary=60000,
    tax_rate=8,
    department="IT",
    employee_id=101,
)
calculate_payroll("Ayesha", 7000, 4000, designation="Backend Developer")
```

```python
def hotel_bill(
    name, *room_rates, service_charge=5, discount=0, tax_rate=10, **booking_details
):

    subtotal = sum(room_rates)

    if discount > 0:
        discount = subtotal * (discount / 100)

    total_bill = subtotal - discount

    service_charge = total_bill * (service_charge / 100)

    total_bill = total_bill + service_charge

    tax = total_bill * (tax_rate / 100)

    final_amount = total_bill + tax

    print("---------Booking Details--------")
    print(f"Name: {name}")
    for key, val in booking_details.items():
        print(f"{key}: {val}")

    print(f"Subtotal: {subtotal}")
    print(f"Discount: {discount}")
    print(f"Service charges: {service_charge}")
    print(f"Tax: {tax}")
    print(f"Final Amount: {final_amount}\n")


hotel_bill(
    "John",
    8000,
    8000,
    10000,
    service_charge=5,
    discount=10,
    tax_rate=12,
    room_type="Deluxe",
    booking_id="HT101",
)
hotel_bill("Sara", 5000, 5000, payment_method="Card", room_type="Standard")
```
