# Python String Slicing – Practice Questions

```python
text = "PythonStringSlicingChallenge"


a = 0
 
for i in range(len(text)):
    print(text[i],a)
    a+=1
```

# 1.	Extract every character except the first 3 and the last 4 characters 

```python
print(text[3:24])
```

# 2.	Extract the last 10 characters using only negative indexes.

```python
print(text[-10:])  
```

# 3.	Extract the substring from index 15 back to index 5 using a negative step.

```python
print(text[15:4:-1]) 
```

# 4.	Extract every second character, starting at index 2 and stopping before index 20.

```python
print(text[2:20:2])
```

# 5.	Use negative indexes to extract every second character from the end of the string toward the beginning


```python
print(text[::-2])
```

# 6.	Extract characters starting at positive index 4 and stopping at negative index -5, taking every third character.

```python
print(text[4:24:3])
```

# 7.	Reverse the substring that starts at index 5 and ends before index 20. Use slicing only.

```python
print(text[19:4:-1])
```

# 8.	Extract the string without its first 5 and last 5 characters, then reverse the result.

```python
print(text[5:23])

print(text[-6:-24:-1])
```

# 9.	Starting from index -3, move backward by 3 positions toward the beginning of the string. Use a negative step.

```python
print(text[-3::-3])
```

# 10.	Extract every second character from the last 20 characters, but in reverse order. Use negative indexes wherever possible.

```python
print(text[:-21:-2])
```
