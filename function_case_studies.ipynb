```python
"""Problem 1: Smart Parking Lot Management
A shopping mall wants to develop a simple parking management system to keep track of vehicles entering and leaving its parking area. The parking lot has a limited capacity of 20 vehicles, and each vehicle is identified by its registration number and vehicle type, such as car, motorcycle, or van. When a vehicle enters, the system should check whether a parking space is available and whether that vehicle is already parked. If space is available, the vehicle should be registered with its entry time; otherwise, the system should inform the driver that the parking lot is full. When a vehicle leaves, the system should remove its registration number from the active parking records and calculate its parking fee according to its vehicle type. The program should continue running until the parking attendant chooses to close it, and separate functions should handle vehicle entry, vehicle exit, and displaying the current parking status.
"""

parking_lot = []
max_capacity = 20
parking_fees = {"car": 10, "motorcycle": 5, "van": 15}


def vehicle_entry(registration_number, vehicle_type):

    for vehicle in parking_lot:
        if vehicle["registration_number"] == registration_number:
            print("This vehicle is already parked")
            return

    entry_time = input("Enter entry time (HH:MM): ")
    parking_lot.append(
        {
            "registration_number": registration_number,
            "vehicle_type": vehicle_type,
            "entry_time": entry_time,
        }
    )
    print(f"Vehicle {registration_number} has been parked")


def vehicle_exit(registration_number):
    count = 0
    for vehicle in parking_lot:
        if vehicle["registration_number"] == registration_number:
            parking_lot.pop(count)
            fee = parking_fees.get(vehicle["vehicle_type"], 0)
            print(
                f"Vehicle {registration_number} has left the parking lot and Parking fee is ${fee}"
            )
            return
        count += 1
    print("Vehicle not found in the parking lot")


def display_parking():
    print("Current parking status:")
    for vehicle in parking_lot:
        print(
            f"Registration: {vehicle['registration_number']}, Type: {vehicle['vehicle_type']}, Entry Time: {vehicle['entry_time']}"
        )


while True:
    print("\nParking Lot Management System")
    print("1. Vehicle Entry")
    print("2. Vehicle Exit")
    print("3. Display Parking Status")
    print("4. Exit System\n")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        if len(parking_lot) >= max_capacity:
            print("Parking lot is full")
        else:
            registration_number = input("Enter vehicle registration number: ")
            vehicle_type = input("Enter vehicle type (car/motorcycle/van): ")
            vehicle_entry(registration_number, vehicle_type)
    elif choice == "2":
        registration_number = input("Enter vehicle registration number: ")
        vehicle_exit(registration_number)
    elif choice == "3":
        display_parking()
    elif choice == "4":
        print("Exiting the system.")
        break
    else:
        print("Invalid choice. Please try again.")
```
