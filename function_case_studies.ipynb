```python
parking_lot = []
max_capacity = 20
parking_fees = {"car": 10, "motorcycle": 5, "van": 15}


def vehicle_entry(registration_number, vehicle_type, entry_time):

    for vehicle in parking_lot:
        if vehicle["registration_number"] == registration_number:
            print("This vehicle is already parked")
            return

    parking_lot.append(
        {
            "registration_number": registration_number,
            "vehicle_type": vehicle_type,
            "entry_time": entry_time,
        }
    )
    print(f"Vehicle {registration_number} has been parked")


def vehicle_exit(registration_number):
    count = 0
    for vehicle in parking_lot:
        if vehicle["registration_number"] == registration_number:
            parking_lot.pop(count)
            fee = parking_fees.get(vehicle["vehicle_type"], 0)
            print(
                f"Vehicle {registration_number} has left the parking lot and Parking fee is ${fee}"
            )
            return
        count += 1
    print("Vehicle not found in the parking lot")


def display_parking_status():
    print("Current parking status:")
    for vehicle in parking_lot:
        print(
            f"Registration: {vehicle['registration_number']}, Type: {vehicle['vehicle_type']}, Entry Time: {vehicle['entry_time']}"
        )


while True:
    print("\nParking Lot Management System")
    print("1. Vehicle Entry")
    print("2. Vehicle Exit")
    print("3. Display Parking Status")
    print("4. Exit System\n")

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        if len(parking_lot) >= max_capacity:
            print("Parking lot is full")
        else:
            registration_number = input("Enter vehicle registration number: ")
            vehicle_type = input("Enter vehicle type (car/motorcycle/van): ")
            entry_time = input("Enter entry time (HH:MM): ")
            vehicle_entry(registration_number, vehicle_type, entry_time)
    elif choice == "2":
        registration_number = input("Enter vehicle registration number: ")
        vehicle_exit(registration_number)
    elif choice == "3":
        display_parking_status()
    elif choice == "4":
        break
    else:
        print("Invalid choice try again")
```

```python
"""Problem 2: Online Quiz Competition
An educational platform is organizing an online quiz competition in which multiple students can participate. Each student must answer a series of questions, and every question has four possible options with only one correct answer. Store the questions, options, and correct answers using appropriate Python data structures. The program should display questions one at a time, accept the student's answer, and award 10 points for every correct answer while deducting 2 points for every incorrect answer. If a student skips a question, no points should be added or deducted. After the quiz ends, display the student's total score, number of correct answers, number of incorrect answers, and number of skipped questions. Allow multiple students to participate and maintain their results so that the program can display the highest-scoring student at the end. Use functions to manage the quiz and calculate results.
"""

QUESTIONS = [
    {
        "question": "Which language is used to write this program?",
        "options": {
            "A": "Python",
            "B": "Java",
            "C": "C++",
            "D": "HTML",
        },
        "answer": "A",
    },
    {
        "question": "Which data structure stores key-value pairs in Python?",
        "options": {
            "A": "List",
            "B": "Tuple",
            "C": "Dictionary",
            "D": "Set",
        },
        "answer": "C",
    },
    {
        "question": "Which keyword defines a function in Python?",
        "options": {
            "A": "function",
            "B": "def",
            "C": "define",
            "D": "fun",
        },
        "answer": "B",
    },
    {
        "question": "What is the result of 2 + 3 * 4?",
        "options": {
            "A": "20",
            "B": "14",
            "C": "24",
            "D": "9",
        },
        "answer": "B",
    },
]


def ask_question(question_number, question):
    print(f"\nQuestion {question_number}: {question['question']}")
    for option, text in question["options"].items():
        print(f"{option}. {text}")

    while True:
        response = input("Your answer (A-D, or press Enter to skip): ").strip().upper()
        if response == "" or response == "SKIP":
            return None
        if response in question["options"]:
            return response
        print("Invalid answer. Please enter A, B, C, D, or SKIP.")


correct_points = 10
incorrect_points = -2


def calculate_result(answers, questions=QUESTIONS):
    correct = 0
    for index in range(len(answers)):
        if answers[index] is not None:
            if answers[index] == questions[index]["answer"]:
                correct += 1
    skipped = 0

    for answer in answers:
        if answer is None:
            skipped += 1
    incorrect = len(answers) - correct - skipped
    score = (correct * correct_points) - (incorrect * 2)

    return {
        "score": score,
        "correct": correct,
        "incorrect": incorrect,
        "skipped": skipped,
    }


def run_quiz(student_name, questions=QUESTIONS):
    print(f"\nWelcome, {student_name}. Answer the following questions.")
    answers = []
    question_number = 1

    for question in questions:
        answer = ask_question(question_number, question)
        answers.append(answer)
        question_number += 1
    result = calculate_result(answers, questions)
    result["name"] = student_name
    return result


def display_result(result):
    print(f"\nResult for {result['name']}")
    print(f"Total score: {result['score']}")
    print(f"Correct answers: {result['correct']}")
    print(f"Incorrect answers: {result['incorrect']}")
    print(f"Skipped questions: {result['skipped']}")


def display_winner(results):
    if not results:
        print("\nNo students participated")
        return

    highest_score = results[0]["score"]
    winners = []

    for result in results:
        if result["score"] > highest_score:
            highest_score = result["score"]
            winners = [result["name"]]
        elif result["score"] == highest_score:
            winners.append(result["name"])
    print(f"\nHighest score: {highest_score}")
    print("Winner(s): " + ", ".join(winners))


results = []
print("Online Quiz Competition")

while True:
    student_name = input("\nEnter student name or 'done' to exit: ").strip()
    if student_name.lower() == "done":
        break
    if not student_name:
        print("Student name cannot be empty.")
        continue

    result = run_quiz(student_name)
    results.append(result)
    display_result(result)

display_winner(results)
```

```python
customers = [
    {
        "account_number": "123",
        "pin": "1234",
        "balance": 1000.0,
        "transaction_history": [],
    },
    {
        "account_number": "456",
        "pin": "5678",
        "balance": 2000.0,
        "transaction_history": [],
    },
]


def authentication(account_number, pin):
    for customer in customers:
        if customer["account_number"] == account_number and customer["pin"] == pin:
            return customer
    return None


def check_balance(customer):
    print(f"Your current balance is: ${customer['balance']:.2f}")


def deposit(customer, amount):
    if amount <= 0:
        print("Deposit amount must be positive.")
        return
    customer["balance"] += amount
    customer["transaction_history"].append(f"Deposited: ${amount:.2f}")
    print(
        f"Successfully deposited ${amount:.2f}. New balance: ${customer['balance']:.2f}"
    )


def withdraw(customer, amount):
    if amount <= 0:
        print("Withdrawal amount must be positive.")
        return
    if amount > customer["balance"]:
        print("Insufficient balance for this withdrawal")
        return
    customer["balance"] -= amount
    customer["transaction_history"].append(f"Withdrew: ${amount:.2f}")
    print(
        f"Successfully withdrew ${amount:.2f}. New balance: ${customer['balance']:.2f}"
    )


while True:
    print("\nATM Transaction Simulator")
    account_number = input("Enter account number (or 'exit' to quit): ").strip()
    if account_number.lower() == "exit":
        break

    pin = input("Enter PIN: ").strip()

    customer = authentication(account_number, pin)
    if customer is None:
        print("Invalid account number or PIN. Please try again.")
        continue
    else:
        print(f"Welcome, account {customer['account_number']}!")

    while True:
        print("\n1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. View Transaction History")
        print("5. Exit\n")

        choice = input("Enter your choice (1-5): ")

        if choice == "1":
            check_balance(customer)
        elif choice == "2":
            amount = float(input("Enter amount to deposit: "))
            deposit(customer, amount)
        elif choice == "3":
            amount = float(input("Enter amount to withdraw: "))
            withdraw(customer, amount)
        elif choice == "4":
            print("\nTransaction History:")
            for transaction in customer["transaction_history"]:
                print(transaction)
        elif choice == "5":
            print(
                f"Thank you for using the ATM, account {customer['account_number']}. Goodbye!"
            )
            break
        else:
            print("Invalid choice. Please try again.")
```
